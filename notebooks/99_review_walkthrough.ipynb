This companion notebook documents the verified U-Net results from the repository and explains what the overlap metrics do—and do not—support.


## 1. Establish the held-out result


```python
# Keep validation and test metrics separate so the selected model is not confused with final evaluation.
print("best_validation_dice: 0.9191")
print("best_validation_loss: 0.1086")
print("test_dice: 0.9115")
print("training_epochs: 20")
```

### What this tells us

The final test Dice score of **0.9115** is close to the best validation Dice of **0.9191**, which is encouraging because the held-out result does not collapse relative to validation. Dice measures mask overlap, so this result supports strong segmentation overlap on the project dataset—not clinical diagnostic accuracy.


## 2. Make the data and architecture scale visible


```python
# The model is trained from scratch rather than relying on a pretrained encoder.
print("dataset_pairs: 10,015")
print("train: 8,012")
print("validation: 1,001")
print("test: 1,002")
print("input_size: 256x256")
print("unet_parameters: 31,037,633")
```

### What this tells us

This is a substantial U-Net trained from scratch on roughly ten thousand image-mask pairs. The train/validation/test separation supports a genuine held-out evaluation, while the absence of a pretrained backbone makes the result a direct test of the U-Net training pipeline rather than transfer learning.


## 3. Explain the objective function


```python
# BCE and Dice address complementary parts of the segmentation problem.
print("loss: 0.5 * BCE + 0.5 * Dice")
print("optimizer: Adam")
print("scheduler: ReduceLROnPlateau")
print("augmentation: flips, rotations, colour jitter")
```

### What this tells us

Binary cross-entropy provides pixel-level supervision while Dice loss directly rewards overlap between predicted and target masks. Combining them is sensible for a lesion/background task where the lesion can occupy a relatively small part of the image.


## Final interpretation

The project demonstrates strong pixel-level overlap on its held-out dermoscopy test set and a complete from-scratch segmentation workflow. It should still be described as a research prototype: Dice overlap does not establish clinical safety, diagnosis quality, or generalization to images collected under different conditions.
